```python
import pandas as pd 
import numpy as np 
import joblib
from sklearn.preprocessing import MinMaxScaler
```

```python
index_names = ['unit_number', 'time_in_cycles']
setting_names = ['op_setting_1', 'op_setting_2', 'op_setting_3']
sensor_names = [f'sensor_{i}' for i in range(1, 22)]
col_names = index_names + setting_names + sensor_names

df = pd.read_csv("../CMAPSSData/train_FD001.txt", sep=r"\s+", header=None, names=col_names)
df.to_csv("../Data/dataset.csv")
print("saved sucessfully")
```

```python
df['max_cycle'] = df.groupby('unit_number')['time_in_cycles'].transform('max')
df['RUL'] = df['max_cycle'] - df['time_in_cycles']
df = df.drop(columns = "max_cycle")
```

```python
#I will be using the learnings from the EDA.ipynb and doing the same as i did there 
cols = ['sensor_1', 'sensor_5', 'sensor_6', 'sensor_10', 'sensor_16', 'sensor_18', 'sensor_19']
df = df.drop(columns = cols)
df = df.drop(columns = setting_names)
df = df.drop(columns = index_names)
```

```python
df.shape
```

```python
# In this i am going to test out a theory where i will be training the 
# auto encoder on just healthy data that is RUL > 50 and check what 
# difference it makes
healthy_df = df[df['RUL'] > 50].drop(columns = ["RUL"])
```

```python
healthy_df
```

```python
def scale():
    x = healthy_df.values
    scaler = MinMaxScaler()
    X = scaler.fit_transform(x)
    return X, scaler 
```

```python
X, scaler = scale() 

np.save("../Data/Train_Data_V2/X_train_v2", X)
joblib.dump(scaler, "../Data/Train_Data_V2/scaler_v2.pkl")
print("FILES SAVED SUCCESSFULLY")
```
